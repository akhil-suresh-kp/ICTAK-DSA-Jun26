##### 3.2 Explain, with a code example in your script, the difference between local and global variable scope in Python.

##### Answer: Local variable is a variable which, when declared in a function, can only be used to print or update its value only inside that particular function, and we cannot print or update that variable outside the function. Global variable on the other hand, when declared outside the function and assigned with a value, that value will only be returned when printed, even though we call the same variable inside the function with a different value.

```python
#example: in the above function apply_discount, the discount_rate and new_dataset are global variables 
#since they are declared outside the function. 

def fun():
    y = 12
    print(y)
fun()
print(y)  

# Here, y is a local variable since its called only inside the function. 
# Printing the value of y outside the fuction will throw an error.
```

##### 4.1 Write a custom iterator class OrderIterator to iterate through the orders, yielding one order at a time.

```python
class OrderIterator:
    def __init__(self, orders):
        self.orders = orders
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < len(self.orders):
            order = self.orders[self.count]
            self.count += 1
            return order
        else:
            raise StopIteration

iterator = OrderIterator(dataset)

for order in iterator:
    print(order)
```

##### 4.2 Create a dictionary grouping all orders by customer using a dictionary comprehension.

```python
grouped_orders = {
    customer: [order for order in dataset if order["customer"] == customer]
    for customer in {order["customer"] for order in dataset}
}
print(grouped_orders)
```
