##### Meaningful Visualisations : Scatter Plot

```python
#Petal Length vs Petal Width
dataset_iris['Classification_num'] = dataset_iris['Classification'].map({
    'Iris-setosa':0,
    'Iris-versicolor':1,
    'Iris-virginica':2
})
```

```python
dataset_iris.plot.scatter(
    x='PL',
    y='PW',
    c='Classification_num',
    cmap='rainbow',
    title='Petal Length vs Petal Width'
)
```
