```python
weather_data = [
    {'location':'kochi','temperature': 30, 'humidity': 95, 'wind': 11, 'air quality': 44, 'precipitation': 15},
    {'location':'mumbai','temperature': 28, 'humidity': 84, 'wind': 15, 'air quality': 60, 'precipitation': 40},
    {'location':'dubai','temperature': 41, 'humidity': 45, 'wind': 18, 'air quality': 95, 'precipitation': 0},
    {'location':'muscat','temperature': 38, 'humidity': 55, 'wind': 16, 'air quality': 70, 'precipitation': 0},
    {'location':'madrid','temperature': 33, 'humidity': 40, 'wind': 13, 'air quality': 38, 'precipitation': 0}
]

five_days_data = [
    {'location':'kochi','temperature':[30,29,31,30,28],'humidity':[95,93,96,94,92]}, 
    {'location':'mumbai','temperature':[28,29,30,29,28],'humidity':[84,86,83,85,82]}, 
    {'location':'dubai','temperature':[41,42,43,42,41],'humidity':[45,43,42,44,46]}, 
    {'location':'muscat','temperature':[38,39,40,39,38],'humidity':[55,53,52,54,56]}, 
    {'location':'madrid','temperature':[33,34,35,34,33],'humidity':[40,42,39,41,43]}
]
```

```python
def weather_predictor(city, weather_data):
    for x in range(len(weather_data)):
        if city == weather_data[x]['location']:
            print("Today's Weather Bulletin: \n")
            print("Humidity (in %):", weather_data[x]['humidity'],)
            print("Rain Amount (in mm):", weather_data[x]['precipitation'])
            if 20 <= weather_data[x]['humidity'] <= 80:
                print("It is expected to be a pleasant sunny day, with possibility of mild clouds")
            else:
                if weather_data[x]['precipitation'] == 0:
                    print("Expect cloudy weather")
                elif 0 < weather_data[x]['precipitation'] <= 5:
                    print("Expect a light rain")
                elif 5 < weather_data[x]['precipitation'] <= 15:
                    print("Expect light to moderate rainfall with occassional downpours")
                elif 15 < weather_data[x]['precipitation'] <= 30:
                    print("Expect moderate to heavy rainfall")
                else:
                    print("Expect heavy downpours. Watch for flooding!")
            

            print("Temperature (in °C):", weather_data[x]['temperature'])
            if weather_data[x]['temperature'] <= 0 or weather_data[x]['temperature'] <= 10:
                print("Very cold to freezing conditions expected. Stay warm.")
            elif 11 <= weather_data[x]['temperature'] <= 20:
                print("Cold climate. Low temperature expected")
            elif 21 <= weather_data[x]['temperature'] <= 30:
                print("Pleasant to warm temperature expected")
            elif 31 <= weather_data[x]['temperature'] <= 35:
                print("Warm climate due to a heatwave that is expected to peak in coming days")
            elif 36 <= weather_data[x]['temperature'] <= 40:
                print("Hot climate due to ongoing heatwave")
            else:
                print("Severe heat climate. Stay hydrated and sheltered as much as possible.")
            

            print("Wind Speed (in km/h):", weather_data[x]['wind'])    
            if weather_data[x]['wind'] < 10 and weather_data[x]['precipitation'] < 5:
                print("Wind will be calm this time")
            elif 10 <= weather_data[x]['wind'] <= 25 and weather_data[x]['precipitation'] <= 15:
                print("Light to moderate winds expected")
            else:
                if weather_data[x]['precipitation'] == 0:
                    print("Expect strong dry winds today, may lead to to potential dust storms at dry areas.")
                else:
                    print("Expect strong winds due to POTENTIAL THUNDERSTORMS")

            print("Air Quality:", weather_data[x]['air quality'])
            if weather_data[x]['air quality'] <= 50:
                print("Good Air Quality")
            elif 51 <= weather_data[x]['air quality'] <= 100:
                print("Satisfiable Air Quality")
            elif 101 <= weather_data[x]['air quality'] <= 150:
                print("Unhealthy for sensitive groups")
            elif 151 <= weather_data[x]['air quality'] <= 200:
                print("Unhealthy")
            elif 201 <= weather_data[x]['air quality'] <= 300:
                print("Very Unhealthy")
            elif 301 <= weather_data[x]['air quality'] <= 500:
                print("Hazardous")

            average_temperature(city,five_days_data)
            average_humidity(city,five_days_data)

        else:
            continue
        


def average_temperature(city,five_days_data):
    for x in range(len(five_days_data)):
        if city == five_days_data[x]['location']:
            total_temp = sum(five_days_data[x]['temperature'])
    print("Average Temperature (in °C) for the next five days:", total_temp/len(five_days_data[x]['temperature']))

def average_humidity(city,five_days_data):
    for x in range(len(five_days_data)):
        if city == five_days_data[x]['location']:
            total_temp = sum(five_days_data[x]['humidity'])
    print("Average Humidity (in %) for the next five days:", total_temp/len(five_days_data[x]['humidity']))   
```
