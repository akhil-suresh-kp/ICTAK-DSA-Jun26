#### BANK ACCOUNT MANAGER

```python
def deposit(balance):
    amount = int(input("Enter the amount to be deposited : "))
    balance += amount
    return balance

def withdrawal(balance):
    amount = int(input("Enter the amount to be withdrawn : "))
    if balance < amount:
        print("Insufficient Balance! Transaction Cancelled")
    else:
        balance -= amount
        print("Withdrawn successfully!")
    return balance

def balance_inquiry(balance):
    print("Balance:", balance)

def transaction(acc_pin,balance):
    op = int(input("SELECT TRANSACTION: 1 for deposit, 2 for withdrawal, 3 for balance inquiry: "))
    if op in [1,2,3]:
        pin = int(input("Enter your PIN: "))
        if pin != acc_pin:
            print("Incorrect PIN! Transaction Cancelled")
        else:
            if op == 1:
                balance = deposit(balance)
                print("Transaction done successfully! Thank you for banking")
            elif op == 2:
                balance = withdrawal(balance)
                
            elif op == 3:
                balance_inquiry(balance)
            
    else:
        print("Invalid option! Transaction cancelled")
    return balance
```
