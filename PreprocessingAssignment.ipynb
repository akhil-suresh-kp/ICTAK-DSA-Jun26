#### Handling Null Values 

```python
data.isna().sum()
```

```python
sel_cols = data[['Sale Price', 'No of Bathrooms', 'Flat Area (in Sqft)', 
            'Lot Area (in Sqft)', 'Area of the House from Basement (in Sqft)', 'Zipcode', 'Latitude', 'Longitude',
            'Living Area after Renovation (in Sqft)']]  #These are the numerical columns with null values
```

##### Note that the column 'No of Times Visited' has the highest number of null values, but it is not included in the sel_cols because this column is an object column. 
##### So we will fill the null values of this object column with its mode. For the sel_cols, we will use simple imputer

##### Lets look at the unique values in 'No of Times Visited'

```python
data['No of Times Visited'].unique()
```

```python
data['No of Times Visited'].mode()  # Twice is the most frequent entry
```

```python
data['No of Times Visited'] = data['No of Times Visited'].fillna(data['No of Times Visited'].mode()[0])
```

#### Outlier Removal

```python
df = pd.read_csv(r'C:\Users\AKHIL SURESH\Downloads\House_Pricing.csv')
```

```python
df.head()
```

```python
sel_cols = df[['Sale Price','No of Bathrooms', 'Flat Area (in Sqft)', 
            'Lot Area (in Sqft)', 'Area of the House from Basement (in Sqft)', 'Zipcode', 'Latitude', 'Longitude',
            'Living Area after Renovation (in Sqft)']]
```

```python
for val in sel_cols:
    df[val] = df[val].fillna(df[val].median())
```

```python
df['No of Times Visited'] = df['No of Times Visited'].fillna(df['No of Times Visited'].mode()[0])
```

```python
df.isna().sum()
```

```python
plt.boxplot(df['No of Bedrooms'])
plt.show()
```

```python
Q1 = np.percentile(df['No of Bedrooms'],25)
Q2 = np.percentile(df['No of Bedrooms'],50)
Q3 = np.percentile(df['No of Bedrooms'],75)
print(Q1,Q2,Q3)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
low_lim = Q1 - 1.5*IQR
upr_lim = Q3 + 1.5*IQR
print(low_lim,upr_lim)
```

```python
outliers = []
for x in df['No of Bedrooms']:
    if x < low_lim or x > upr_lim:
        outliers.append(x)

print(outliers)
```
